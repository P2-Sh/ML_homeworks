```python
import numpy as np

# 1. Батч: 10 объектов, 4 признака
X = np.random.randn(10, 4)

# 2. Скрытый слой: 6 нейронов
W1 = np.random.randn(4, 6)
b1 = np.random.randn(6)

# 3. Линейный слой
z = X @ W1 + b1

# 4. ReLU
h = np.maximum(0, z)

# 5. Выходной слой: 3 класса
W2 = np.random.randn(6, 3)
b2 = np.random.randn(3)

logits = h @ W2 + b2

# 6. Softmax
shifted = logits - np.max(logits, axis=1, keepdims=True)
exp_logits = np.exp(shifted)
probs = exp_logits / np.sum(exp_logits, axis=1, keepdims=True)

# 7. Правильные классы для 10 объектов
y = np.array([0, 2, 1, 0, 2, 1, 0, 1, 2, 0])

# 8. Cross-Entropy
correct_probs = probs[np.arange(len(y)), y]
loss = -np.mean(np.log(correct_probs))

print("X:", X.shape)
print("z:", z.shape)
print("h:", h.shape)
print("logits:", logits.shape)
print("probs:", probs.shape)
print("Cross-Entropy:", loss)

# Проверка Softmax
print("Суммы вероятностей:", probs.sum(axis=1))
```
